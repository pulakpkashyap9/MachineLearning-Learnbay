## Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:,0:3], df['Sales'], test_size = 0.3, random_state = 5)
```

```python
x_train
```

```python
x_test
```

## Standardizing the Input Variables

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
print(x_train.shape)
x_train
```

```python
x_test = scaler.transform(x_test)
print(x_test.shape)
x_test
```

## Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_reg = LinearRegression()
linear_reg.fit(x_train, y_train)
```

```python
print(linear_reg.intercept_) # Results are based on OLS
```

```python
print(linear_reg.coef_) # Results are based on OLS
```

```python
import statsmodels.api as sm
```

```python
# Here, we will have to add intercept
x_train_sm = sm.add_constant(x_train)
```

```python
lr1 = sm.OLS(y_train, x_train_sm).fit()
```

```python
lr1.params
```

```python
pred = linear_reg.predict(x_test)
pred
```

```python
np.array(y_test)
```

## Evaluation

```python
lr1.summary()
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error, mean_absolute_percentage_error
```

```python
mse = mean_squared_error(y_test, pred)
mae = mean_absolute_error(y_test, pred)
rmse = root_mean_squared_error(y_test, pred)
mape = mean_absolute_percentage_error(y_test, pred)
print(f'Mean Squared Error is: {mse}')
print(f'Mean Absolute Error is: {mae}')
print(f'Root of the Mean Squared Error is: {rmse}')
print(f'Mean Absolute Percentage Error is: {mape}')
```
